# class 상속
- 부모(슈퍼)클래스의 코드를 자식(서브) 클래스가 물려받는 것

## 부모 클래스

```python
class Animal: # 부모 클래스
    def __init__(self): # 생성자
        self.height = 30 # 인스턴스 변수

    def get_height(self): # 인스턴스 매서드
        print(f'Animal {self.height}')
```

```python
fubao = Animal()
fubao.get_height()
```

## 자식 클래스

```python
class Dog(Animal):
    pass

class Cat(Animal):
    pass
```

```python
bow = Dog()
bow.get_height()
mow = Cat()
mow.get_height()
```

```python
class Dog(Animal):
    def cry(self):
        print('멍멍')

class Cat(Animal):
    def run(self):
        print('사뿐사뿐')
```

```python
bow = Dog()
bow.get_height()
bow.cry()
mow = Cat()
mow.get_height()
mow.run()
```

```python
class Parent:
    def __init__(self, name):
        self.name = name

    def hello(self):
        print(f'안녕하세요 {self.name}입니다.')
```

```python
class child(Parent):
    def bye(self):
        print(f'{self.name}은/는 떠납니다.')
```

```python
person1 = Parent('홍길동')
person1.hello()

child1 = child('홍이동')
child1.hello()
child1.bye()
```

```python
class child(Parent):
    def __init__(self, name, age):
        super().__init__(name)
        self.age = age

    def bye(self):
        print(f'{self.age}살 {self.name}은/는 떠납니다.')
    
```

```python
child2 = child('홍삼동', 20)
child2.bye()
```

```python
class child(Parent):
    def __init__(self, name, age):
        super().__init__(name)
        self.age = age

    def bye(self):
        super().hello()
        print(f'{self.age}살 {self.name}은/는 떠납니다.')
```
