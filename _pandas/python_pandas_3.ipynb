## 판다스 데이터 가공하기

```python
import pandas as pd
import numpy as np

filename = 'dust1.xlsx'
```

```python
dust = pd.read_excel(filename)
dust.head(5)
```

## 컬럼(열, 변수, 피쳐) 삭제 / 생성

### drop(컬럼명, aixs=1)--> 컬럼(열) 삭제
- drop(행, aixs=0)--> 행 삭제 (aixs=0은 생략 가능)

```python
# 지역, 망, 측정소코드 삭제
dust2 = dust.drop(['지역','망','측정소코드'], axis=1)
dust2.head(5)
```

```python
dust.head(5)
```

### 컬럼 추가

- 데이터프레임명['추가하고싶은 컬럼 이름'] = 값

```python
dust['city'] = '서울'
```

```python
dust.head()
```

## 컬럼 이름 수정 --> rename()

- 데이터프레임명.rename(columns={'기존 이름':'새 이름'})

```python
dust2 = dust.rename(columns={'측정소명':'name',
                     '측정일시':'date',
                     '주소':'addr'})
```

```python
dust2.columns
```

## 데이터 형 확인 --> dtypes

```python
dust.dtypes # 판다스 3버전부터 object -> str 변경
```

## 데이터 형 변환 --> astype()

- 컬럼명.astype(바꿀 자료형)

```python
dust2['date'].astype(str)
```

```python
dust2['date'] = dust2['date'].astype(str)
```

```python
dust2.dtypes
```

```python
dust2['date']
```

```python
# 첫 8자리만 자른다 --> slice
dust2['date'] = dust2['date'].str.slice(0,8)
```

# 시간 날짜 형식으로 바꾸기 --> to_datetime()

```python
dust2['date'] = pd.to_datetime(dust2['date'])
dust2['date']
```
