## Matplotlib basic 기본 시각화

```python
# 라이브러리 불러오기
import matplotlib.pyplot as plt
```

```python
# 데이터
x_data = [2,3,4,5]
y_data = [1,7,5,6]

plt.plot(x_data,y_data) # 꺾은선 그래프
plt.title('title') # 차트 제목
```

## 범례 --> plt.legend()

```python
import koreanize_matplotlib

data1 = [2,3,4,5]
data2 = [8,6,4,2]
plt.plot(data1, label='오름차순')
plt.plot(data2, label='내림차순')
plt.legend()
```

## 선 색 바꾸기

```python
plt.plot(x_data,y_data, color='red')
```

## x축 제목, y축 제목

```python
plt.plot(x_data,y_data)
plt.xlabel('x축')
plt.ylabel('y축')
```

## 선 종류
- linestyle='dash' or linestyle='--'

```python
plt.plot(x_data, linestyle='--')
plt.plot(y_data, color='g', linestyle=':')
plt.plot(data2, color='g', linestyle='-.')
plt.xlabel('x축')
plt.ylabel('y축')
```

## 마커 종류

```python
plt.title('x, y range')
xdata = [5,10,15,20]
ydata = [1,3,5,7]
plt.plot(xdata, ydata, marker='o')
```

## 축 범위 조정

```python
plt.title('x, y range')
xdata = [5,10,15,20]
ydata = [1,3,5,7]
plt.plot(xdata, ydata, marker='o')
plt.xlim(0,25) # x 축 범위 지정
plt.ylim(-2,10) # y 축 범위 지정
```
