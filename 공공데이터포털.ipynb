## 서울시 의류수거함 위치 현황

```python
import csv
position = set()
x = []
y = []
with open('서울특별시 관악구_의류수거함위치_20210121.csv','r') as f:
    r = csv.reader(f)
    for i,c in enumerate(r):
        if i == 0:
            continue
        position.add(c[1])
        x.append(float(c[2]))
        y.append(float(c[3]))
```

```python
# 위도 경도 정규화
n_x=[]
n_y=[]

for xx in x:
    n = xx/(max(x)-min(x))
    n_x.append(n)
    
for yy in y:
    n = yy/(max(y)-min(y))
    n_y.append(n)
```

```python
import matplotlib.pyplot as plt

plt.scatter(x,y)
plt.show()
```
