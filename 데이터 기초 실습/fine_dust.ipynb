## 에어코리ㅏ 미세먼지 데이터

# fine_dust.xlsx 엑셀파일, 라이브러리

```python
import pandas as pd
import matplotlib.pylab as plt

data = pd.read_excel('fine_dust.xlsx', index_col='area')
data.head()
```

## 2018년 지역별 미세먼지 선 그래프

```python
data2018 = data[2018]
data2018
```

```python
import koreanize_matplotlib

plt.figure(figsize=(15,4))
plt.plot(data2018, color='b', marker='o')
plt.title('2018년 미세먼지 그래프')
plt.grid() # 격자
```

## 2016~ 2019 지역별 미세먼지 선 그래프 

```python
for y in range(2016,2020):
    data1619 = data[y]
    print(data1619)
```

```python
import matplotlib.pyplot as plt
import koreanize_matplotlib
plt.figure(figsize=(15,4))
for y in range(2016,2020):
    data1619 = data[y]
    plt.plot(data1619, marker='s', label=y)
plt.legend()
plt.title('2016-2019 지역별 미세먼지')
plt.xlabel('지역')
plt.ylabel('미세먼지량')
plt.grid()
plt.show()
```

## 2017 지역별 미세먼지 세로 막대 그래프
- plt.bar(x,높이(value))

```python
data2017 = data[2017]
# data2017
data.index
```

```python
plt.figure(figsize=(15,4))

plt.bar(data.index, data2017, color='g')
plt.grid()
plt.title('2017년 지역별')
plt.xlabel('지역')
plt.ylabel('micrometer')
plt.show()
```

## 2016-2019 특정지역 미세먼지

```python
plt.figure(figsize=(15,4))

# 경기 인천 부산 대구
data4 = data.loc['Gyeonggi':'Daegu',2016:2019]
```

```python
for y in range(2016,2020):
    chartd = data4[y]
    plt.bar(data4.index, chartd)
```

```python
import numpy as np
index = np.arange(4)

for y in range(2016,2020):
    chartd = data4[y]
    plt.bar(index, chartd, width=0.2, label=y)
    index = index+0.2

plt.ylim(35,55)
plt.xlabel('지역')
plt.ylabel('마이크로미터')
plt.legend()
plt.xticks(index-0.5 ,['경기','인천','부산','대구']) # x축 눈금 재설정
plt.show()
```

```python
for y in range(2016,2020):
    chartd = data4[y]
    plt.barh(data4.index, chartd)
plt.xlim(30,55)
```
