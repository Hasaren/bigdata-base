# 파일 입출력

## 1. 파일 객체 = open() ---> 마지막에 파일 객체명.close()

### 파일열기 모드
|종류|설명|특징|
|----|----|-----|
|w|쓰기 모드 (write)|파일이 없는 경우 새로 생성, 있으면 덮어쓰기|
|a|추가 모드 (append)|파일이 없는 경우 새로 생성, 있으면 내용 추가|
|r|읽기 모드 (read)|파일이 없는 경우 에러, 읽어서 처리|

### 파일의 종류
|종류|설명|특징|파일|
|----|----|----|----|
|t|텍스트 파일|생략 가능|txt, csv, docx, ...|
|b|바이너리 파일|생략 불가|그림, 소리, 동영상 등|

```python
f = open('./test.txt','w',encoding='utf-8')
f.write("안녕")
f.close()
```

```python
f = open('./test.txt','w',encoding='utf-8')
while True:
    name = input()
    if not name:
        break
    f.write(name)
    f.write('\n')
f.close()
```

```python
f = open('./test.txt','r',encoding='utf-8')
f.readlines()
```

## 2. with open() as 파일 객체명:
|종류|설명|특징|
|----|----|-----|
|w|쓰기 모드 (write)|파일이 없는 경우 새로 생성, 있으면 덮어쓰기|
|a|추가 모드 (append)|파일이 없는 경우 새로 생성, 있으면 내용 추가|
|r|읽기 모드 (read)|파일이 없는 경우 에러, 읽어서 처리|

```python
with open('./test.txt') as f:
    a = f.readlines()
    print(a)
```

## 실습 예제 - 오늘의 스케줄 입력하기

```python
import time

with open(time.strftime('%Y-%m-%d')+'.txt', 'w', encoding='utf-8') as f:
    f.write('안녕')
```

# 번역기 만들기

```python
with open('공지사항.txt', 'w', encoding='utf-8') as f:
    f.write('안녕')
```

```python
from googletrans import Translator as tl

trans = tl()
with open('공지사항.txt', encoding='utf-8') as f:
    notice = f.read()
    print('원문')
    print('-'*80)
    print(notice)
    print('-'*80)

    code = ['en','ja','zh-TW'] #영어, 일본어, 중국어

    while True:
        choice = int(input("1영어, 2일본어, 3중국어, 0종료: "))
        if choice == 0:
            break
        lang = code[choice-1]
        res = trans.translate(notice, src = 'ko', dest=lang)
        print(res.text)
```
