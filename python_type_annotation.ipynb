# type annotation
- 3.5버전 이상에서 사용가능
- 강제성이 없는 자료형에 관한 힌트를 알려준다.

## 기존 변수 선언 방법

```python
num = 1
li = [1,2,3,4]
d = {'name':'푸바오', 'age':5}
print(num, li, d, sep='\n')
```

```python
print(type(num),type(li),type(d))
```

## annotation을 넣을 수 있는 변수 선언 방법

```python
# 변수이름: 자료형 = 값
num: int = 1
li : list = [1,2,3,4]
d: dict = {'name':'푸바오', 'age':5}
```

## 일반적인 함수 정의

```python
def add(a,b):
    return a+b
```

```python
print(add(1,2))
print(type(add(1,2)))
```

```python
print(add(1.2,2.2))
print(type(add(1.2,2.2)))
```

```python
print(add('안녕','하세요'))
print(type(add('안녕','하세요')))
```

```python
print(add([1,2],[3,4]))
print(type(add([1,2],[3,4])))
```

## annotation 변수 함수 정의

```python
# def 함수명(매개변수: 자료형) -> 반환형의 자료형:
def sub(a: int, b: int) -> int:
    return a-b
```

```python
print(sub(1,2))
print(type(sub(1,2)))
```

```python
print(sub(1.2,2.2))
print(type(sub(1.2,2.2)))
```
