## loc

- 데이터프레임명.loc[행, 열]
- 열번호는 생략 가능하나 행번호는 생략 불가

```python
df.head()
```

```python
df.loc[0,['date','temp']]
```

```python
df.loc[0] # 열은 생략 가능
```

```python
df.loc[:,'temp'] # :의 의미는 모든 행
```

```python
df.loc[:,['date','temp']] # 열 여러개 가져올 경우
```

```python
df.loc[0:5] # loc는 마지막 번호까지 포함한다. (range와 다름)
```

## iloc

- i는 정수 의미
- 데이터 프레임명.iloc[행번호, 열번호]
- 열번호는 생략 가능하나 행번호는 생략 불가

```python
df.iloc[0,1]
```

```python
df.iloc[1:3] # 마지막 번호 포함 안함 (range와 같음)
```

```python
df.iloc[0,0:2]
```

```python
df.iloc[1:3,0:3]
```

# 판다스 결측데이터 처리

## isna() --> 값이 nan(결측치)이라면 True

```python
# 최대풍속이 측정되지 않은 모든 데이터 출력
df[df['max_wind'].isna()]
```

## 데이터프레임명.isnull().sum() --> 컬럼별 결측데이터 개수 확인

```python
df.isnull().sum()
```

```python
df.isnull().sum().sum() #전체 결측치 수
```

## 결측치 삭제하기 --> dropna()

```python
df.dropna() # 결측치를 제거한 상태를 확인만 하는 용도
```

```python
df # 원본 데이터 유지되어 있다.
```

```python
df2 = df.dropna()
df2
```

```python
df2.isnull().sum()
```

## 결측 데이터 대체하기 --> fillna(채울 값)

```python
df['max_wind'].isnull().sum()
df3 = df
```

```python
df3['max_wind'] = df3['max_wind'].fillna(0)
df3['max_wind'].isnull().sum()
```

```python
avg = df3['max_wind'].mean()
avg
```

```python
df3['max_wind'] = df3['max_wind'].fillna(avg)
df3['max_wind'].isnull().sum()
```

```python
avg2 = df3['mean_wind'].mean()
df3['mean_wind'] = df3['mean_wind'].fillna(avg2)
```
