```python
import pandas as pd
import numpy as np

dust = pd.read_excel('dust1.xlsx')
dust.head()
```

```python
# 문지열 형 변환
dust['측정일시'] = dust['측정일시'].astype(str)
dust['측정일시'] = dust['측정일시'].str.slice(0,8)
```

```python
# 날짜형으로 변환
dust['측정일시'] = pd.to_datetime(dust['측정일시'])
dust['측정일시']
```

## 날짜형을 년, 월, 일로 분리 추출

```python
# dt 속성
dust['측정일시'].dt
```

```python
dust['측정일시'].dt.year
```

## 년, 월, 일 컬럼 생성

```python
dust['year'] = dust['측정일시'].dt.year # 연도 추출
dust['month'] = dust['측정일시'].dt.month # 월 추출
dust['day'] = dust['측정일시'].dt.day # 일 추출
```

```python
dust.head()
```

# 데이터 병합


```python
s1 = pd.read_excel('nation.xlsx')
s2 = pd.read_excel('code.xlsx')
```

```python
s1.head()
```

```python
s2.head()
```

## pd.merge(데이터프레임1, 데이터프레임2, on=기준칼럼명, how='inner')

- how='inner' 'outer','left','right' 도 가능 (교집합, 차집합, 합집합 등)

```python
# 국적코드 기준으로 merge
pd.merge(s1,s2,on='국적코드') 
```

```python
# 오른쪽 컬럼 기준 (차집합)
pd.merge(s1, s2, how='right', on='국적코드')
```

# 데이터 그룹핑

## groupby()

```python
# 국적코드 별 성별, 입국객수, 증가수
s1.groupby('국적코드').sum()
```

```python
# 국적코드 별, 성별 별 ---> 입국객 수, 증가수
s1.groupby(['국적코드','성별']).sum()
```
