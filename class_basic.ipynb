# 클래스와 객체지향 프로그래밍

## 클래스란 무엇인가

**클래스(class)** 는 데이터와 그 데이터를 다루는 함수(메서드)를 하나로 묶어 새로운 타입을 정의하는 방법 (설게도)

| 용어 | 설명 | 비유 |
|------|------|------|
| class | 설계도 | 모형 틀 |
| 인스턴스/객체 | 클래스로 만든 실제 객체 | 틀로 찍어낸 것 |
| 메서드 | 클래스 내부에서 만들어진 함수 | 객체의 기능 |
| 속성 | 객체가 가진 데이터 | 객체의 세부 속성 |

## 클래스 기본 구조

```python
# 클래스 정의의 기본 구조
# class 클래스 이름:
# 파이썬에서 클래스의 이름 첫글자를 대문자로 하는 것이 관행
class Toy:
    # 메서드 정의
    # self : 현재 만들어진 객체 자기 자신을 가리킨다.
    def __init__(self): # 생성자
        self.elems = [] # self.속성 --> 객체 변수

    # 객체 메서드
    def add(self, new_elems):
        """
        첫번째 매개변수 반드시 self --> 호출한 인스턴스가 자동으로 전달된다.
        new_elems : 리스트이다.
        """
        self.elems+=new_elems
    
    def size(self):
        """
        리스트 elems의 길이를 출력
        """
        print(len(self.elems))
```

## 클래스로 객체 만들기

> 인스턴스 이름 = 클래스이름()

```python
t1 = Toy()
t2 = Toy()
```

## 메서드 호출

```python
t1.add([3,4])
```

```python
t1.size()
```

## 클래스 예제

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def print_info(self):
        print(f'{self.name}은 {self.age}세 입니다.')
```

```python
p1 = Person('홍길동',30)
p1.print_info()
```

```python
# 속성 메서드 추가

class Person:
    def __init__(self, name, age, hobby, job):
        self.name = name
        self.age = age
        self.hobby = hobby
        self.job = job

        self.print_age()
        self.print_hobby()
        self.print_job()
    
    def print_age(self):
        print(f'{self.name}의 나이: {self.age}')
        
    def print_hobby(self):
        print(f'{self.name}의 취미: {self.hobby}')
        
    def print_job(self):
        print(f'{self.name}의 직업: {self.job}')
        
```

```python
p1 = Person('홍길동',20,'요리','요리사')
```
