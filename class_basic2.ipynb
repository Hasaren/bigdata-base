# 클래스와 클래스 변수

- 여러 인스턴스가 변수공간을 같이 사용하는 변수
- 같은 클래스를 통해서 만들어진 인스턴스끼리 공간을 사용

```python
class Person:
    population = 0

    def __init__(self, name):
        """
        생성자
        """
        Person.population += 1
        self.name = name

    def print_hobby(self, hobby):
        print(hobby)
```

```python
p1 = Person('홍길동')
p1.print_hobby('요리')
```

```python
p1.population
```

## 클래스 매서드

- 인스턴스가 아니라 클래스 자체를 대상으로 동작
- 클래스 변수 수정
- 클래스 전체 설정 변경
- @classmethod 라는 데코레이터 사용
- 첫번째 매개변수로 cls사용
- 클래스 매서드는 객체를 만들지 않고 바로 호출 가능

### 사용 이유

- 여러 객체가 있을 대 공통 데이터를 관리하기 위함

```python
class Student:
    school = '학교' # 클래스 변수

    def instance_method(self):
        print("인스턴스 메서드")

    @classmethod
    def class_method(cls):
        print('클래스 매서드')
```

```python
Student.class_method()

s1 = Student()
s1.instance_method()
```

```python
class Person:
    population = 0

    def __init__(self, name):
        """
        생성자
        """
        Person.population += 1
        self.name = name

    def print_hobby(self, hobby):
        print(hobby)

    @classmethod
    def get_popul(cls):
        return cls.population
```

```python
p1 = Person('ㄱ')
p2 = Person('ㄴ')

Person.get_popul()
```

## 실습 예제
- 분식집 -> 메뉴, 개수 -> 총금액

```python
class Bunsik:
    total = 0
    menu_d = {'김밥':3000, '떡볶이': 6000}
    def __init__(self, menu, cnt):
        self.menu = menu
        self.cnt = cnt
        Bunsik.total += Bunsik.menu_d[menu]*cnt
        
    @classmethod
    def print_total(cls):
        print(cls.total)
```

```python
Bunsik('김밥',2)
```

```python
Bunsik('떡볶이',2)
```

```python
Bunsik.print_total()
```
